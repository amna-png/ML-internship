**Decision:** The primary decision improved by this work is identifying which high-impression web pages the content team should review first for a title/meta rewrite due to underperformance.

**Action:** A content writer or editor acts on this recommendation by taking the top-ranked candidates and editing their titles and meta descriptions.

**Cost of a false positive:** A writer spends time rewriting a page whose low CTR was already normal for its query intent. This is a one-time, bounded cost.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

visible = df[(df["avg_position"] > 0) & (df["impressions_90d"] >= 500)]
print("Total rows:", len(df))
print("Visible, high-impression pool:", len(visible))

tier_ctr = visible.groupby("position_tier")["ctr"].agg(["mean", "median", "count"])
print(tier_ctr.sort_values("mean", ascending=False))

tier_median = visible.groupby("position_tier")["ctr"].transform("median")
gap_pool = visible[visible["ctr"] < 0.5 * tier_median]
print("Pages below half their tier's median CTR:", len(gap_pool), "of", len(visible))
```
